##  Comment fonctionne RFECV ?

1. **Applique RFE** :  
   - Il entraîne un modèle et élimine les features les moins importantes, une par une.

2. **Evalue chaque étape avec la validation croisée** :  
   - À chaque itération, il teste la performance du modèle avec un nombre réduit de features.  
   - Il utilise la **validation croisée (CV)** pour éviter le surajustement.

3. **Sélectionne le meilleur nombre de features** :  
   - Il choisit la configuration où le modèle a la meilleure performance.

```python
from sklearn.feature_selection import RFECV
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedKFold
import pandas as pd

# Charger les données (exemple avec un dataset fictif)
df = pd.read_csv("dataset.csv")

# APRES le pré-traitement:
# Définir X (features) et y (variable cible)
X = df.drop(columns=["target"])  # Suppression de la colonne cible
y = df["target"]  # Définition de la variable cible

# Modèle de base
model = LogisticRegression()

# RFECV avec validation croisée (5 folds)
rfecv = RFECV(estimator=model, cv=StratifiedKFold(5))

# Entraînement sur les données préalablement coupées
rfecv.fit(X, y)

# Nombre optimal de features sélectionnées
print("Nombre optimal de variables :", rfecv.n_features_)

# Liste des features retenues (True = gardée, False = éliminée)
print(rfecv.support_)

# Rang des features (1 = la plus importante)
print(rfecv.ranking_)
```
